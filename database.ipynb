# MySQL Insert

```python
import mysql.connector

mydb = mysql.connector.connect(
  host="database-1.c21fjnvngsvf.us-east-1.rds.amazonaws.com",
  user="admin",
  password="Senai12345",
  database="mydatabase"
)

mycursor = mydb.cursor()

sql = "INSERT INTO customers (name, address) VALUES (%s, %s)"
val = [
  ("Ana", "Rua das Flores 123"),
  ("Carlos", "Av. Brasil 456"),
  ("Mariana", "Rua do Sol 789"),
  ("Pedro", "Travessa das Palmeiras 101"),
  ("Juliana", "Av. Central 202")
]

mycursor.executemany(sql, val)

mydb.commit()

print(mycursor.rowcount, "records inserted.")
```

# MySQL Select

```python
mycursor.execute("SELECT * FROM customers")
myresult = mycursor.fetchone()
print(myresult)
```

# MySQL Where

```python
sql = "SELECT * FROM customers WHERE address = 'Rua das Flores 123'"
mycursor.execute(sql)
myresult = mycursor.fetchall()
for x in myresult:
  print(x)
```

# MySQL Order By

```python
sql = "SELECT * FROM customers ORDER BY name"
mycursor.execute(sql)
myresult = mycursor.fetchall()
for x in myresult:
  print(x)
```

# MySQL Update

```python
sql = "UPDATE customers SET address = 'Canyon 123' WHERE address = 'Rua das Flores 123'"
mycursor.execute(sql)
mydb.commit()
```

# MySQL Delete

```python
sql = "DELETE FROM customers WHERE address = 'Av. Central 202'"
mycursor.execute(sql)
mydb.commit()
```

# MySQL Limit

```python
mycursor.execute("SELECT * FROM customers LIMIT 5")
myresult = mycursor.fetchall()
for x in myresult:
  print(x)
```

# MySQL Join

```python
sql = '''
  SELECT
    customers.name AS user,
    customers.address AS favorite_address
  FROM customers
'''
mycursor.execute(sql)
myresult = mycursor.fetchall()
for x in myresult:
  print(x)
```
